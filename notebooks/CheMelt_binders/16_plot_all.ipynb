```python
from pychemelt.utils.plotting import (
    plot_unfolding, 
    PlotConfig, 
    LegendConfig,
    AxisConfig,
)

import pickle
import os
```

```python
dir = "./py_objects/"
pickles = os.listdir(dir)
pickles = [dir + p for p in pickles if p.endswith(".pkl")]
# Sort them from 0_ to 14_
pickles.sort(key=lambda x: int(os.path.basename(x).split("_")[0]))

names = [os.path.basename(p).split(".")[0].split('_')[1] for p in pickles]

class Test:
    pass 

objs = {}

# Open them
for i,p in enumerate(pickles):
    with open(p, "rb") as f:
        t = pickle.load(f)
        objs[names[i]] = t
```

```python
# Create legend and plot configs
plot_config = PlotConfig()
plot_config.font_size = 40

legend_config = LegendConfig()
legend_config.color_bar_orientation = "v"
legend_config.color_bar_length = 0.5
legend_config.color_bar_x_pos = 0.7
legend_config.color_bar_y_pos = 0.82

axis_config = AxisConfig()
```

```python
# Plot all of them 
for i in range(len(objs)):
    obj = objs[names[i]]
    name = names[i]
    plot_name = f"./figures/{name}_fits_scaled.png"
    print('Plotting:', plot_name)
    
    fig = plot_unfolding(
        obj,
        plot_config=plot_config,
        legend_config=legend_config,
        axis_config=axis_config,
        use_scaled_data=True)
    
    fig.write_image(plot_name)
```

```python
# Create a Figure that includes the thirteen fits
fig_list = os.listdir('./figures/')
fig_list = [x for x in fig_list if x.endswith('.png') and 'fits_scaled' in x]
# Remove P6A from the list and also G5A
fig_list = [x for x in fig_list if 'P6A' not in x]
fig_list = [x for x in fig_list if 'G5A' not in x]
fig_list = ['./figures/' + x for x in fig_list]  # Prepend the directory path to each filename
```

```python

from PIL import Image, ImageDraw, ImageFont
import math
import string
import os

# Open images
images = [Image.open(f).convert("RGB") for f in fig_list]
panel_titles = [
    os.path.splitext(os.path.basename(f))[0].split('_')[0]
    for f in fig_list
]

w, h = images[0].size

cols = 3
rows = math.ceil(len(images) / cols)
title_pad = 60

combined = Image.new("RGB", (cols * w, rows * (h + title_pad)), "white")

from matplotlib import font_manager

title_font_path = font_manager.findfont(
    font_manager.FontProperties(family="DejaVu Sans", weight="bold")
)
title_font = ImageFont.truetype(title_font_path, 32)

label_font_path = font_manager.findfont(
    font_manager.FontProperties(family="DejaVu Sans", weight="bold")
)
label_font = ImageFont.truetype(label_font_path, 40)

draw = ImageDraw.Draw(combined)

# Paste images and add filename-derived titles
for i, (img, title) in enumerate(zip(images, panel_titles)):
    tile_x = (i % cols) * w
    tile_y = (i // cols) * (h + title_pad)
    img_y = tile_y + title_pad

    combined.paste(img, (tile_x, img_y))

    # Center title above each subplot
    title_bbox = draw.textbbox((0, 0), title, font=title_font)
    title_w = title_bbox[2] - title_bbox[0]
    title_h = title_bbox[3] - title_bbox[1]
    title_x = tile_x + (w - title_w) // 2
    title_y = tile_y + (title_pad - title_h) // 2
    draw.text((title_x, title_y), title, fill="black", font=title_font)

# Add A, B, C... labels on each subplot
for i in range(len(images)):
    label = string.ascii_uppercase[i]

    tile_x = (i % cols) * w
    tile_y = (i // cols) * (h + title_pad)
    img_y = tile_y + title_pad

    x = tile_x + 20
    y = img_y + 20

    # White rectangle behind the text
    bbox = draw.textbbox((x, y), label, font=label_font)
    draw.rectangle(
        (bbox[0] - 8, bbox[1] - 8, bbox[2] + 8, bbox[3] + 8),
        fill="white"
    )

    draw.text((x, y), label, fill="black", font=label_font)

combined.save("./figures/fits_all.png")
```
